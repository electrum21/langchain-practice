## Tool definition

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.tools import tool

@tool
def square_root(x: float) -> float:
    """Calculate the square root of a number"""
    return x ** 0.5
```

```python
@tool("square_root")
def tool1(x: float) -> float:
    """Calculate the square root of a number"""
    return x ** 0.5
```

```python
@tool("square_root", description="Calculate the square root of a number")
def tool1(x: float) -> float:
    return x ** 0.5
```

```python
tool1.invoke({"x": 467})
```

Learning Points:
- Remember to add @tool decorator
- Provide a intuitive description for the agent to understand
- The agent will invoke the tool by using the .invoke() method

## Adding to agents

```python
from langchain.agents import create_agent

agent = create_agent(
    model="google_genai:gemini-3.1-flash-lite",
    tools=[tool1],
    system_prompt="You are an arithmetic wizard. Use your tools to calculate the square root and square of any number."
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="What is the square root of 467?")

response = agent.invoke(
    {"messages": [question]}
)

print(response['messages'][-1].content[0]['text'])
```

```python
from pprint import pprint

pprint(response['messages'])
```

```python
print(response["messages"][1].tool_calls)
```

Learning Points:
- The response from the tool back to the agent is called a tool message
- The tool message provides the output from the tool call (which in this case is the answer)
