```python
from langchain.chat_models import init_chat_model

model = init_chat_model(model="google_genai:gemini-3.1-flash-lite")
```

## Customising your Model

```python
model = init_chat_model(
    model="google_genai:gemini-3.1-flash-lite",
    # Kwargs passed to the model:
    temperature=1.0 
)

response = model.invoke("What's the capital of the Moon?")
print(response.content[0]['text'])
```

Learning Points: 
- Higher temperature makes output more creative, lower temperature makes output more deterministic

## Initialising and invoking an agent

```python
from langchain.agents import create_agent

agent = create_agent(model=model)
```

```python
agent = create_agent(model="claude-sonnet-4-6")
```

```python
agent = create_agent("google_genai:gemini-3.1-flash-lite")
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?")]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
print(response['messages'][-1].content[0]['text'])
```

```python
from langchain.messages import AIMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
) 

pprint(response)
```

Learning Points:
- HumanMessage signifies that it is calling from the user
- Response is a dictionary of messages
- To access the answer, take the last message and the content block within it
- Can pass in a chat history with several human messages and AI messages back and forth
- There was latency observed, the output is produced at one go after a delay of certain duration

## Streaming Output

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]},
    stream_mode="messages"
):

    # token is a message chunk with token content
    # metadata contains which node produced the token
    
    if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```

Learning Points:
- By using streaming, the output becomes gradually produced instead of the whole chunk being displayed at one go
- This is similar to using an AI Chatbot, where the responses are streaming in as time passes by
